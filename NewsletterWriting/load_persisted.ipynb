# Load persisted clustering outputs

`load_persisted` reads three S3 datasets for one job date and joins them for the writers:
- **articles** from the enrichment bucket
- **clusters** (metrics + cheap summaries) from `news-clusters--eun1-az1--x-s3`
- **article ↔ cluster ids** from `articles-cluster-ids--eun1-az1--x-s3`

```python
import sys
from pathlib import Path

src = Path(".").resolve() / "src"
if str(src) not in sys.path:
    sys.path.insert(0, str(src))

from editorial.config import EditorialConfig
from editorial.s3 import load_persisted

# Set a date like "2026-09-13", or None for UTC today.
config = EditorialConfig(job_date=None)
loaded = load_persisted(config)

print(loaded["job_date"])
print(len(loaded["articles"]), "articles")
print(len(loaded["clusters"]), "clusters")
loaded["clusters"][0] if loaded["clusters"] else None
```
